```python
if 'gmail' in df.columns:
    df['gmail'] = df['gmail'].str.replace('\xa0', ' ') 

if 'password' in df.columns:
    df['password'] = df['password'].str.replace('\xa0', ' ')

print(df)
```

```python
gaccounts = df["gmail"]
gpasswords = df["password"]
```

```python
sender_gaccount = ''
sender_gpass = ''
smtp_server = 'smtp.gmail.com'
smtp_port = 587
```

```python
msg_subject = 'Routine security procedures'
msg_body =""" 
Hello,
I am Abdullah El Saudi
This is a security routine to protect the Gmails from being blocked.
"""

message =f'Subject: {msg_subject}\n\n{msg_body}'
```

```python
# Configure logging
log.basicConfig(filename='Gmail_log.txt', level=log.INFO, 
                    format='%(asctime)s - %(levelname)s - %(message)s')

for sender_gaccount, sender_gpass in zip(gaccounts, gpasswords):
    try:
        with smtplib.SMTP(smtp_server, smtp_port) as smtp:
            log.info(f'Sending from {str(sender_gaccount)}')
            smtp.starttls()
            smtp.login(sender_gaccount, sender_gpass)
            smtp.sendmail(sender_gaccount, gaccounts, message)
            log.info(f'Send done at {datetime.datetime.now()}')
            log.info(f'Waiting {waittime} seconds')
            time.sleep(waittime)
            print(f"sending from {sender_gaccount}")
            print('Done')
            print(f"waiting {waittime} seconds")
    except smtplib.SMTPAuthenticationError:
        log.error(f"Failed to authenticate {sender_gaccount}. Check your username/password.")
    except smtplib.SMTPException as e:
        log.error(f"SMTP error occurred: {e}")
    except Exception as e:
        log.error(f"An error occurred: {e}")

print("All tasks completed.")
```
